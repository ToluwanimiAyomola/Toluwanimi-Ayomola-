```python

"""
Student Result Analyzer
By: Ayomola Toluwanimi Henry
Date: Oct 2026
A simple Python project to analyze student results
"""

import pandas as pd

# Sample Student Data
students = {
    'Name': ['Ayomola', 'David', 'Grace', 'Henry', 'Sarah'],
    'Math': [85, 78, 92, 65, 88],
    'English': [90, 82, 88, 70, 95],
    'Science': [88, 75, 95, 68, 90]
}

# Create DataFrame
df = pd.DataFrame(students)

# Calculate Total and Average
df['Total'] = df['Math'] + df['English'] + df['Science']
df['Average'] = df['Total'] / 3
df['Grade'] = df['Average'].apply(lambda x: 'A' if x>=80 else 'B' if x>=60 else 'C')

# Display Results
print("=== STUDENT RESULT ANALYZER ===")
print(df)
print("\n=== TOP PERFORMER ===")
print(df.loc[df['Average'].idxmax()])

print("\n=== CLASS STATISTICS ===")
print(f"Class Average: {df['Average'].mean():.2f}")
print(f"Highest Score: {df['Average'].max():.2f}")
print(f"Lowest Score: {df['Average'].min():.2f}")

# Save to CSV
df.to_csv('results.csv', index=False)
print("\n✅ Results saved to results.csv")
```
