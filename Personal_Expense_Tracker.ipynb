```python

import pandas as pd

expenses = {
    "Date": [
        "2026-10-15",
        "2026-10-15",
        "2026-10-16",
        "2026-10-16",
        "2026-10-17",
        "2026-10-17",
        "2026-10-18"
    ],
    "Category": [
        "Food",
        "Transport",
        "Data",
        "Food",
        "Education",
        "Transport",
        "Food"
    ],
    "Description": [
        "Lunch",
        "Bus fare",
        "Internet subscription",
        "Dinner",
        "Learning materials",
        "Bus fare",
        "Breakfast"
    ],
    "Amount_NGN": [
        1500,
        500,
        2000,
        1800,
        3000,
        700,
        1200
    ]
}

df = pd.DataFrame(expenses)

df
```

```python

total_expenses = df["Amount_NGN"].sum()

print("PERSONAL EXPENSE TRACKER")
print("------------------------")
print(f"Total Expenses: ₦{total_expenses:,.2f}")
print(f"Number of Transactions: {len(df)}")
print(f"Average Expense: ₦{df['Amount_NGN'].mean():,.2f}")
```

```python

category_expenses = df.groupby("Category")["Amount_NGN"].sum()

print("EXPENSES BY CATEGORY")
print("====================")

print(category_expenses)

print("\nHighest Spending Category:")
print(category_expenses.idxmax())

print(f"\nHighest Amount Spent: ₦{category_expenses.max():,.2f}")
```

```python

category_expenses = df.groupby("Category")["Amount_NGN"].sum()

category_expenses.plot(kind="bar")

plt.title("Personal Expenses by Category")
plt.xlabel("Expense Category")
plt.ylabel("Amount Spent (₦)")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

category_expenses = df.groupby("Category")["Amount_NGN"].sum()

plt.figure(figsize=(8, 8))

plt.pie(
category_expenses,
labels=category_expenses.index,
autopct="%1.1f%%",
startangle=90
)

plt.title("Percentage of Expenses by Category")
plt.tight_layout()
plt.show()
```

```python

df.to_csv("personal_expenses.csv", index=False)

print("Dataset saved successfully!")
print("Filename: personal_expenses.csv")
```
